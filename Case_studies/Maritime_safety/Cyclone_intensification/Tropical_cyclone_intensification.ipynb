### Outline

Hurricanes are hitting Central and North American coasts every summer / early fall. Such meteorological events are fueled by the energy (heat) of the ocean. Intensification, or at the reverse intensity decrease of theses hurricanes (and of typhoons and cyclones) can occur in relation with oceanic structures such as currents or eddies. Late September / early October 2024, two such hurricanes hit South-East United States of America (in particular Florida, but also northernmost States), Helene and Milton. Comparing the hurricanes' intensity (in term of wind speed or of categories) with the sea level anomalies or with a quantity which can be computed using sea level anomales and in situ data (or using models), the Tropical Cyclone Heat Potential, a data generated and disseminated by NOAA, can give hints on a possible intensification.

In the case of Helene and Milton, such comparisons close to their landing on US coasts could suggest that it may have intensified the former, slightly decreased the latter (which ended up being the strongest-ever hurricane in the area anyway). A large part of this notebook is using the notebook on cyclones from altimetry, with a different focus and different data comparisons and overlays.

## Defining functions

</div>

We are going to use a number of quick functions in this notebook. There calculate TCHP and support our plotting, but we are going to hide them by default.

```python
def compute_tchp_fast(ds, temp_var='thetao', depth_dim='depth', rho=1025.0, cp=3985.0):
    """
    Compute Tropical Cyclone Heat Potential (TCHP) quickly, without interpolation

    Parameters
    ----------
    ds : xr.Dataset
        Dataset containing temperature
    temp_var : str
        Temperature variable name (default: 'thetao')
    depth_dim : str
        Depth dimension name (default: 'depth')
    rho : float
        Seawater density (kg/m^3)
    cp : float
        Heat capacity (J/kg/K)

    Returns
    -------
    xr.DataArray
        TCHP in kJ/cm^2 with dims (time, lat, lon)
    """

    T = ds[temp_var]
    z = ds[depth_dim].values

    # --- layer thickness (1D) ---
    dz = np.diff(z)
    dz = np.append(dz, dz[-1])
    dz = np.abs(dz)

    dz_da = xr.DataArray(dz, dims=[depth_dim])

    # --- heat content above 26°C ---
    T_excess = (T - 26.0).clip(min=0)

    tchp = (T_excess * dz_da).sum(dim=depth_dim)

    # --- convert to energy (kJ/cm^2) ---
    tchp = rho * cp * tchp / 1e7

    # --- tidy output order ---
    dims_out = [d for d in ['time', 'latitude', 'longitude', 'lat', 'lon'] if d in tchp.dims]
    tchp = tchp.transpose(*dims_out, ...)

    tchp.name = "TCHP"
    tchp.attrs["units"] = "kJ cm^-2"
    tchp.attrs["description"] = "Tropical Cyclone Heat Potential"

    return tchp
```

```python
def plot_gridded_data(m, ds, bg_data, storm_date):
    """
    Quick function to plot background L4 gridded data
    """
    
    if bg_data == "gridSLA":
        field_label = "CMEMS gridded sea level anomaly [cm]"
        ds_day = ds.sel(time=storm_date, method='nearest')
        bg = m.contourf(ds_day.longitude, ds_day.latitude, ds_day.sla, levels=np.linspace(-0.15,0.35,128), vmin=-0.15, vmax=0.35, 
                   cmap=background_cmap, extend='both', transform=cartopy.crs.PlateCarree(), zorder=3)
        listeticks = np.linspace(-0.15,0.35,6)
    elif bg_data == "TCHP_NOAA" or "TCHP_CMEMS":
        field_label = "Tropical cyclone heat potential [$kJcm^{-2}$]"
        ds_day = ds.sel(time=storm_date, method='nearest')
        bg = m.contourf(ds_day.longitude, ds_day.latitude, ds_day.Tropical_Cyclone_Heat_Potential, levels=np.linspace(0,120,128), 
                    vmin=0, vmax=120, cmap=background_cmap, extend='both')
        listeticks = np.linspace(0,120,6)
        
    return bg, listeticks, field_label
```

```python
def plot_hurricane_track(m, hurricane_cmap, lon, lat, radius, plot_extents, xoffset, yoffset):
    """
    Quick function to plot a hurricane track
    """
    norm = mpl.colors.Normalize(0,5)
    edgecols = hurricane_cmap(norm(cat))
    sm = plt.cm.ScalarMappable(cmap=hurricane_cmap, norm=norm)
    sm.set_array(edgecols)
    
    track1 = m.scatter(lon[0:count+1], lat[0:count+1], radius[0:count+1], facecolors='None', edgecolors=edgecols, 
                      alpha=0.75, zorder=4, linewidths=rad_edgewidth)

    m.set_extent(plot_extents, crs=cartopy.crs.PlateCarree())

    if lon[count] + xoffset/4 > plot_extents[1]:
        plot_lon = lon[count] - xoffset/4
    else:
        plot_lon = lon[count] + xoffset/10

    if lat[count] + yoffset/4 > plot_extents[3]:
        plot_lat = lat[count] - yoffset/4
    else:
        plot_lat = lat[count] + yoffset/10

    return sm, plot_lat, plot_lon
```

```python
def plot_altimetry_tracks(m, storm_date, time_window, granule_dates, product_dirs):
    """
    Quick function to plot altimetry tracks
    """
    tmin = storm_date - datetime.timedelta(hours=time_window[0])
    tmax = storm_date + datetime.timedelta(hours=time_window[1])
    ii = list(set(np.where(granule_dates >= tmin.to_pydatetime())[0]).intersection(np.where(granule_dates < tmax.to_pydatetime())[0]))

    for product_dir, granule_date in zip(product_dirs[ii], granule_dates[ii]):
        print(product_dir)
        # read data
        if "SEN3" in product_dir:
            # Sentinel-3 SRAL branch (no netCDF groups)
            ds3 = xr.open_dataset(os.path.join(product_dir, 'standard_measurement.nc'))
            plot_var = np.array(ds3.swh_ocean_01_ku)
            flags = np.array(ds3.swh_ocean_qual_01_ku).astype(float) + np.array(ds3.surf_type_01).astype(float)
            alon = ds3["lon_01"]
            alat = ds3["lat_01"]
            ds3.close()
        else:
            # Sentinel-6 branch (with netCDF groups)
            input_file = glob.glob(os.path.join(product_dir, '*.nc'))[0]
            ds6 = xr.open_dataset(input_file, group="data_01", decode_times=False)
            dsc = xr.open_dataset(input_file, group="data_01/ku", decode_times=False)
            plot_var = np.array(dsc.swh_ocean)
            flags = np.array(dsc.swh_ocean_qual).astype(float)+np.array(ds6.surface_classification_flag).astype(float)
            alon = ds6['longitude']
            alat = ds6['latitude']
            alon[flags != 0.0] = np.nan
            alat[flags != 0.0] = np.nan
            m.scatter(alon, alat, c='0.5', s=200, zorder=6)
            ds6.close()
            dsc.close()
        
        plot_var[flags != 0.0] = np.nan
        p_alt = m.scatter(alon, alat, c=plot_var, s=100,
                          cmap=track_cmap, marker='o', edgecolors=None, 
                          linewidth=0.0, vmin=0.0, vmax=7.0, zorder=7)
        return p_alt
```

```python
def make_annotations(m, cat, lon, lat, plot_lon, plot_lat, storm_date):
    """
    Quick function to make plot annotations
    """
    m.annotate(cat[count], xy=(lon[count], lat[count]), xycoords='data',
               xytext=(plot_lon, plot_lat), textcoords='data',
               arrowprops=dict(arrowstyle="->", connectionstyle="arc3"),
               color='k', zorder=100, fontsize=14)
    m.annotate(str(storm_date.date())+' '+str(storm_date.time()), (0.00, -0.0775), xycoords='axes fraction')
```

Now we have the file, lets open it and take a look at its contents using *xarray*...

```python
tracks = xr.open_dataset(local_tracks_file, engine="netcdf4")
tracks
```

We need to find the storm that corresponds to the name and year selection criteria we specified above. In order to do that we will find all instances of our name (**name_ix**) and cross reference with our storm year (**year_ix**).

```python
# find the correct storm for the correct year
name_ix = np.where(np.array(tracks["name"]).astype(str) == storm_name)
storm_dates = np.array(tracks["iso_time"])[name_ix].astype(str)
storm_years = np.array([str(i[0][0:4]) for i in storm_dates])
year_ix = np.where(storm_years == storm_year)[0]

# get relevant coordinates
lon = np.array(tracks["lon"])[name_ix][year_ix]
lat = np.array(tracks["lat"])[name_ix][year_ix]
date = np.array(tracks["iso_time"])[name_ix][year_ix].astype(str)
cat = np.array(tracks["usa_sshs"])[name_ix][year_ix] # Saffir-Simpson scale categories. 
radius = np.array(tracks["usa_roci"])[name_ix][year_ix]
tracks.close()
```

In some cases, when we read in the data we get empty values. We will quickly tidy up our storm record to remove these. 

```python
# strip empties
ii = np.where(date != '')
lon = lon[ii]
lat = lat[ii]
cat = cat[ii]
date = date[ii]
radius = radius[ii]
```

We will also convert our radius values in nautical miles to approximate map pixels using the scale factors we set above. We will use the Saffir-Simpson categories defined for hurricanes (cat 1 to 5), and put all other numbers to zero (for the color scale)

- tropical_depression ≤ 34 knots (≤ 17 m/s  / ≤ 63 km/h)
- tropical_storm: 35-63 knots (18-32 m/s / 63-118 km/h)
- hurricane cat 1: 64-82 knots (33-42 m/s / 119-153 km/h)
- hurricane cat 2: 83-95 knots (43-49 m/s / 154-177 km/h)
- hurricane cat 3: 96-112 knots (50-58 m/s / 178-208 km/h)
- hurricane cat 4: 113-136 knots (58-70 m/s / 209-251 km/h)
- hurricane cat 5: ≥ 137 knots (≥ 70 m/s / 251 km/h )

```python
radius = radius*rad_scale - rad_edgewidth*2
cat = np.maximum(cat, 0) # only keeping "hurricane" categories (1-5 ; NB. the data begins at -3)
```

To help us compare the dates from the storm record with those of the altimetry products we will use the *pandas* library to convert the storm dates from their native datetime64 format.

```python
# format data in pandas
storm_dates = pandas.to_datetime(date)
```

## Acquiring tropical cyclone heat potential products from NOAA ERDAP or the CMEMS Data Store

</div>

Tropical Cyclone Heat Potential (or TCHP) is an indicator of the energy a storm can draw from the ocean and become a tropical cyclone, or intensify when already a cyclone. NOAA compute it using altimetry and in situ data. Another way to compute it can be to use an ocean model providing temperature at depths. We take this latter approach with the CMEMS product.

```python
if bg_data == "TCHP_NOAA":
    local_tchp_file = os.path.join(os.getcwd(), f"TCHP_NOAA_{storm_name}.nc")
    if os.path.exists(local_tchp_file):
        os.remove(local_tchp_file)
    remote_tchp_file = f'https://cwcgom.aoml.noaa.gov/erddap/griddap/aomlTCHP.nc?Tropical_Cyclone_Heat_Potential%5B({start.strftime("%Y-%m-%dT%H:%M:%S")}):1:({end.strftime("%Y-%m-%dT%H:%M:%S")})%5D%5B({plot_extents[2]}):1:({plot_extents[3]})%5D%5B({plot_extents[0]}):1:({plot_extents[1]})%5D,D26%5B({start.strftime("%Y-%m-%dT%H:%M:%S")}):1:({end.strftime("%Y-%m-%dT%H:%M:%S")})%5D%5B(15.0):1:(32.0)%5D%5B(-100.0):1:(-80.0)%5D'
    zipfile = local_tchp_file.replace('.nc','.gz')

    # get the file
    with requests.get(remote_tchp_file, stream=True) as r:
        with open(zipfile, 'wb') as f:
            shutil.copyfileobj(r.raw, f)

    # unzip it
    with gzip.open(zipfile, 'rb') as f_in:
        with open(local_tchp_file, 'wb') as f_out:
            shutil.copyfileobj(f_in, f_out)

    # tidy up
    if os.path.exists(zipfile):
        os.remove(zipfile)
        
    ds = xr.open_dataset(local_tchp_file)
    
elif bg_data == "TCHP_CMEMS":
    local_tchp_file = f"TCHP_CMEMS_{storm_name}.nc"
    if os.path.exists(local_tchp_file):
        os.remove(local_tchp_file)
        
    ds = copernicusmarine.open_dataset(
       dataset_id="cmems_mod_glo_phy-thetao_anfc_0.083deg_P1D-m",
       minimum_longitude = plot_extents[0],
       maximum_longitude = plot_extents[1],
       minimum_latitude = plot_extents[2],
       maximum_latitude = plot_extents[3],
       start_datetime = "2024-10-03T18:00:00",
       end_datetime = "2024-10-12T18:00:00",
       variables = ['thetao'],
       minimum_depth=0,
       maximum_depth=150,
    )
    ds.to_netcdf(local_tchp_file)
    ds = xr.open_dataset(local_tchp_file)
```

In the case of the CMEMS product, we need to calculate the TCHP ourselves. We start with the 4D potential temperature fields an integrate the heat above the 26$^{o}$C isotherm.

```python
if bg_data == "TCHP_CMEMS":
    thetao = ds['thetao'].chunk({'depth': -1, 'time': 1})
    tchp = compute_tchp_fast(ds.assign(thetao=thetao))
    ds["Tropical_Cyclone_Heat_Potential"] = tchp
```

## Gathering altimetry products

</div>

Now that all of our altimetry data is downloaded, we need to collect what we have. The cell below will find all of our Sentinel-3 and Sentinel-6 products.

```python
# find the products
product_dirs_S3 = glob.glob(os.path.join(download_dir, "*.SEN3"))
product_dirs_S6 = glob.glob(os.path.join(download_dir, "*.SEN6"))
```

We will now loop through these products to find the average acquisition time of each product, using the timestamps in the filename.

```python
# loop through granules to get their average time stamps
granule_dates_S3 = []
for product_dir in product_dirs_S3:
    t_step = datetime.datetime.strptime(os.path.basename(product_dir).split('_')[-14], "%Y%m%dT%H%M%S") \
           - datetime.datetime.strptime(os.path.basename(product_dir).split('_')[-15], "%Y%m%dT%H%M%S")
    granule_dates_S3.append(datetime.datetime.strptime(os.path.basename(product_dir).split('_')[-15], "%Y%m%dT%H%M%S") \
                         + t_step/2)
```

Lets ingest the Sentinel-6 data in exactly the same way.

```python
# loop through granules to get their average time stamps
granule_dates_S6 = []
for product_dir in product_dirs_S6:
    t_step = datetime.datetime.strptime(os.path.basename(product_dir).split('_')[-11], "%Y%m%dT%H%M%S") \
           - datetime.datetime.strptime(os.path.basename(product_dir).split('_')[-12], "%Y%m%dT%H%M%S")
    granule_dates_S6.append(datetime.datetime.strptime(os.path.basename(product_dir).split('_')[-12], "%Y%m%dT%H%M%S") \
                         + t_step/2)
```

Finally, lets gather all the Sentinel-3 SRAL and Sentinel-6 products, and their dates, together into a single array.

```python
product_dirs = np.array(product_dirs_S3 + product_dirs_S6)
granule_dates = np.array(granule_dates_S3 + granule_dates_S6)
```

## Plotting Significant Wave Height images

</div>

It has taken us a while to get here, but we are finally ready to launch our plotting script. We will create a series of images that we convert into an animation at the end. Firstly, we will create an output directory for our images...

```python
# Create an output directory for our images products
output_dir = os.path.join(os.getcwd(), f"output_{storm_name}")
os.makedirs(output_dir, exist_ok=True)
```

And now we are ready to go. The following cell looks very complex, but we can't break it up as it runs in a loop that iterates over the storm times. However, it actually performs only a few simple (if specific) tasks;
* *Step 1: Sets up the figure panel for this iteration, and plots the land layer*
* *Step 2: plots the specified ocean grid layer*
* *Step 3: Plots the storm track*
* *Step 4: Finds and relevant altimetry tracks that are in scope for this time slice and plots the significant wave height*
* *Step 5: Annotate the plots*
* *Step 6: Embellish the plot with gridlines etc.*
* *Step 7: Make the colour bars for the storm category (from the IBTrACS record) significant wave height from the altimetry tracks and ocean grid layer*
* *Step 8: Close and figure ready for the next iteration*

Each step is marked in the cell, and calls on of the functions we defined above. We will get an update on progress every "panel_frequency" times steps. 

...so lets go!

```python
# make the animation frames
for storm_date, count in zip(storm_dates, range(len(storm_dates))):

    if np.mod(count, panel_frequency) == 0:
        print(f"Processing data for {storm_date} (panel {str(count).zfill(2)})")
    else:
        continue
        
    # limit to plot region
    if lon[count] < np.nanmin(plot_extents[0]) or lon[count] > np.nanmax(plot_extents[1]) \
      or lat[count] < np.nanmin(plot_extents[2]) or lat[count] > np.nanmax(plot_extents[3]):
        continue

    # Step 1: set up the figure
    fig, [m, m2] = plt.subplots(2, 1, figsize=(plot_xy[0], plot_xy[1]), dpi=150, gridspec_kw={'height_ratios': [14, 1]},
                                subplot_kw={"projection": cartopy.crs.PlateCarree()})
    m2.set_visible(False)
    m.add_feature(land_50m, edgecolor='k', facecolor='k', zorder=2)

    # Step 2: Plot gridded data    
    bg, listeticks, field_label = plot_gridded_data(m, ds, bg_data, storm_date)

    # Step 3: plot the historical hurricane track and latest position -- use Saffir-Simpson categories for color map (0 cat 1-5 Hurricane)
    sm, plot_lat, plot_lon = plot_hurricane_track(m, hurricane_cmap, lon, lat, radius, plot_extents, xoffset, yoffset)

    # Step 4: find and plot the altimetry tracks in this time window
    p_alt = plot_altimetry_tracks(m, storm_date, time_window, granule_dates, product_dirs)
    
    # Step 5: make annotations
    make_annotations(m, cat, lon, lat, plot_lon, plot_lat, storm_date)
    
    # Step 6: embellish this plot
    make_embellishments(m)

    # Step 7: make colour bars
    make_colourbars(m, p_alt, sm, bg, listeticks, field_label)

    # Step 8: save and close figure
    plt.savefig(os.path.join(output_dir, f"panel{str(count).zfill(4)}.png"), bbox_inches='tight')
    plt.close(fig)
```
